```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats, integrate
from math import comb, perm, factorial, log, exp

rng = np.random.default_rng(42)   # graine : memes resultats a chaque execution
```

## Fonctions utiles

```python
def var_tvar_discrete(valeurs, probas, alpha):
    valeurs, probas = np.asarray(valeurs, float), np.asarray(probas, float)
    F = np.cumsum(probas); i = np.argmax(F >= alpha - 1e-12); v = valeurs[i]; m = valeurs > v
    tvar = (np.sum(valeurs[m]*probas[m]) + v*(F[i] - alpha)) / (1 - alpha)
    cte = np.sum(valeurs[m]*probas[m]) / np.sum(probas[m]) if probas[m].sum() > 0 else v
    return v, tvar, cte

def kupiec(x, n, p):
    ph = x/n; l0 = (n - x)*np.log(1 - p) + x*np.log(p)
    l1 = (n - x)*np.log(1 - ph) + (x*np.log(ph) if x > 0 else 0)
    LR = -2*(l0 - l1); return LR, stats.chi2.sf(LR, 1)
z = stats.norm.ppf
```
